```python
Piloto = []

lospilotos = []
print("==================================================")
print("  🤖 IA DE CONTROL DE TRIPULACIÓN INICIADA 🤖   ")
print("==================================================")
print("Estado actual: No hay pilotos registrados en el sistema.")

while True:
    print("\n================ PANÉL DE CONTROL ================")
    print("1. Registrar nuevo piloto")
    print("2. Mostrar todos los pilotos")
    print("3. IA: Evaluar estado de la tripulación")
    print("4. Apagar IA y salir")
    print("==================================================")

    opcion = input("Seleccione una opción (1-4): ")

    if opcion == "1":
        print("\n--- REGISTRO DE NUEVO PILOTO ---")
        piloto_nuevo = {}

        piloto_nuevo["nombre"] = input("Nombre del piloto: ")
        piloto_nuevo["horas_vuelo"] = int(input("Horas de vuelo acumuladas: "))
        piloto_nuevo["estado"] = input("¿Su estado de salud es 'Optimo' o 'Fatigado'?: ")

        lospilotos.append(piloto_nuevo)
        print(f"✅ ¡IA: Piloto {piloto_nuevo['nombre']} registrado con éxito!")

    elif opcion == "2":
        print("\n--- BITÁCORA DE PILOTOS REGISTRADOS ---")
        if lospilotos == []:
            print("🤖 IA: La lista 'lospilotos' está vacía. Registre a alguien primero.")
        else:
            contador = 1
            for p in lospilotos:
                print(f"Piloto N°{contador}: {p['nombre']} | Horas: {p['hours_vuelo'] if 'hours_vuelo' in p else p['horas_vuelo']} | Estado: {p['estado']}")
                contador = contador + 1

    elif opcion == "3":
        print("\n--- ANÁLISIS DE LA IA DE VUELO ---")
        if lospilotos == []:
            print("🤖 IA: Alerta. No puedo evaluar el vuelo porque no hay pilotos en la lista.")
        else:
            primer_piloto = lospilotos[0]
            print(f"🤖 IA: Analizando al Capitán al mando: {primer_piloto['nombre']}...")

            if primer_piloto["estado"] == "Fatigado":
                print("🚨 ALERTA DE LA IA: El piloto está fatigado. ¡Peligro de accidente! Se sugiere relevo.")
            elif primer_piloto["horas_vuelo"] < 50:
                print("⚠️ ADVERTENCIA DE LA IA: El piloto tiene pocas horas de experiencia. Activando copiloto automático.")
            else:
                print("💚 REPORTE DE LA IA: Piloto altamente calificado y en condiciones. ¡Buen viaje!")

    elif opcion == "4":
        print("\n🤖 IA: Guardando datos en 'lospilotos'...")
        print("Sistemas apagados. ¡Hasta la próxima, comandantes!")
        break

    else:
        print("❌ Opción no válida. Intente con un número del 1 al 4.")
print("Programa realizado por Oscar Flores ML = 22")
```
