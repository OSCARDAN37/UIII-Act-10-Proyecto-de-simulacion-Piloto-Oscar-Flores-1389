```python
lospilotos = []

piloto = {}

print("====================================================")
print("¡Bienvenido al Sistema de IA para Escuderías!")
print("====================================================")

while True:
    print("\n--- MENÚ DE OPCIONES ---")
    print("1. Registrar nuevo piloto (Entrenar a la IA)")
    print("2. Ver todos los pilotos registrados")
    print("3. Consultar IA (Predicción de Categoría)")
    print("4. Salir del programa")

    opcion = input("\nSelecciona una opción (1-4): ")

    if opcion == "1":
        print("\n--- Registro de Piloto ---")
        nombre = input("Nombre del piloto: ")
        edad = int(input("Edad del piloto: "))
        experiencia = int(input("Años de experiencia en karts/carreras: "))

        piloto = {
            "nombre": nombre,
            "edad": edad,
            "experiencia": experiencia
        }

        lospilotos.append(piloto)
        print(f"¡Sistema alimentado con éxito! {nombre} ha sido registrado.")

    elif opcion == "2":
        print("\n--- Base de Datos de Pilotos ---")

        if lospilotos == []:
            print("La base de datos está vacía. ¡Registra a alguien primero!")
        else:
            for p in lospilotos:
                print(f"Piloto: {p['nombre']} | Edad: {p['edad']} años | Experiencia: {p['experiencia']} años")


    elif opcion == "3":
        print("\n--- Consulta a la IA de la Escudería ---")

        if lospilotos == []:
            print("La IA no tiene datos para analizar. Registra pilotos primero.")
        else:
            nombre_buscar = input("¿De qué piloto quieres que la IA prediga su categoría?: ")
            encontrado = False

            for p in lospilotos:
                if p["nombre"] == nombre_buscar:
                    encontrado = True
                    print(f"\n[Analizando datos de {p['nombre']}...]")

                    if p["edad"] < 18 and p["experiencia"] < 3:
                        print("Predicción de la IA: El piloto debe ir a la 'Categoría Junior de Desarrollo'.")
                    elif p["edad"] < 18 and p["experiencia"] >= 3:
                        print("Predicción de la IA: ¡Talento Joven! Recomendado para 'Fórmula de Promoción'.")
                    elif p["edad"] >= 18 and p["experiencia"] < 5:
                        print("Predicción de la IA: Recomendado para 'Categoría GT Aficionado'.")
                    elif p["edad"] >= 18 and p["experiencia"] >= 5:
                        print("Predicción de la IA: ¡Piloto Experto! Listo para la 'Categoría Pro / Fórmula 1'.")
                    else:
                        print("Predicción de la IA: Datos inusuales. Se sugiere más pruebas en simulador.")

            if encontrado == False:
                print("El piloto no está registrado en el sistema.")

    elif opcion == "4":
        print("\nApagando los motores del sistema de IA... ¡Adiós!")
        break

    else:
        print("\n[Error] Opción no válida. Elige un número del 1 al 4.")
```
